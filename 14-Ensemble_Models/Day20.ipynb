### Boosting

#### What is Boosting?

Consider a dataset with multiple records

Create Base learners sequentially

Create a Base Learner BL1

Choose the records that are incorrectly classified with the first base learner and pass them to the next Base Learner BL2

BL2 will be trained mostly on the incorrect records so that BL2 can correct them, this time some other records might be incorrectly classified

Choose the records which are incorrectly classified by BL2 and pass them to BL3

Repeat the above process

Specify the number of Base Learners to be defined

### AdaBoost

#### Base Learners in Adaboost are Stumps

Stumps = Decision Trees with depth 1

Consider a dataset with m records and n independent features

Create a column called Sample weight and initialize it as 1/m (where m is the number of records)

What should be the root node of the stump? Decide it based on the Entropy value

Create Stumps with each column as a root node (f1, f2, ..... , fn) and calculate the entropy and choose the best base learner

The one with less entropy is chosen as best base learner model

Calculate total error for incorrect classified records by summing up all the incorrectly classified records sample weight

TE = sum(incorrect records sample weights)

**Performance of Stump** = (1/2) * log ((1-TE)/TE)  where TE = Total Error

Now update the weights based on the Performance

For **incorrectly** classified records:

New sample weight = old sample weight * e^(Performance)

For **correctly** classified records:

New sample weight = old sample weight * e^(-Performance)

Normalize the new weights such that the sum of all updated sample weights results to 1

**normalized_weights** = updated_weights/sum(updated_weights)

Now divide the normalized weigths into buckets

Iterate over this dataset for m times and choose a record randomly in each iteration

Now compare the weight of that randomly selected record in the buckets and check in which bucket range it is falling

Once obtained populate that record in the new dataset and repeat the same with all other records

Now a new dataset is obtained and most of the times we observe the records that are ranged in the incorrectly classified bucket

And repeat the entire process with this new dataset until no records are incorrectly classified

#### Combining weak learners (stumps) to make a strong learner (Adaboost)

Let's say 3 Stumps (base learners) are created for a dataset while training

How to get the final O/P for the test data?

Identify the O/P of each Stump and choose the **majority O/P** for classification & mean/median for regression

### Gradient Boosting

Consider having a dataset with regression data (Eg. Experience, Degree, Salary)

Experience & Degree are independent features, Salary is a dependent feature

Compute the base model which will result an O/P which is equal to the average of dependent variable

Initial y_pred = mean(y)

Compute Residuals (errors / pseudo residuals)

Use a simple loss function (Eg: 1/2*(y_actual - y_pred) ** 2) (Note: Loss function must be derivable / differentiable)

Residual1 (R1) = y-y_pred

Now construct a Decision Tree (Base model 2) which takes same independent features (Experience & Degree) but O/P as Residual1 for training

Now you get y_pred2, again calculate the Residual2 (R2) = R1-y_pred2

Repeat it (create a new DT (Base model 3)) and compute next residuals

Note: Residual keeps decreasing as we progress

Final O/P = (Base model 1 O/P) + alpha1 * (Base model 2 O/P) + alpha2 * (Base model 3 O/P) + ...

Final O/P = h0(x) + alpha1*h1(x) + alpha2*h2(x) + alpha3*h3(x) + ...

alpha_i is a hyperparameter where i = 1,2,3....

### XGBoost (Extreme Gradient Boosting)

#### Classification

Consider a dataset as follows

Features: Salary(<=50K,>50K) & CreditScore(Bad/Normal/Good) (independent), Approval(0/1) (dependent)

Classification problem (0/1)

For Base model1, the probability to calculate the residuals is 0.5 (P = 0.5)

Initial y_pred = mean(0,1) = 0.5

Residual1 R1 = y - y_pred

**Note**: In XG Boost always choose a Binary Classifier Decision Tree as the base model (even if it has >2 categories)

So initiate a base model2 binary classifier DT

Calculate the similarity weights & Information Gain

**Similarity Weight** = sum(residuals)/(sum(P(1-P))+Lambda) where P = 0.5 initially, but will be changed in coming steps

Gain = (similarity_weight_leftNode + similarity_weight_rightNode - similarity_weight_rootNode)

Note: The base model will be the one with the max Gain

Calculate Gain by taking different independent features as rootNode & choose the one with the max Gain

Once the rootNode, left & rightNodes of depth1 are created, extend the childNodes with respect to remaining independent features

Again choose them based on highest Gain basis only

How depth can we go?

Post Pruning => **Cover Value**

Let's say if the Gain is less than the Cover Value, then stop splitting that node

Thus, Base Model 1 is created now

Note: BaseModel_i O/P = log(odds) = log(P/1-P)

Thus initial Base Model's O/P = log(0.5/0.5) = log(1) = 0

**New P** or Final O/P = sigmoid(Base Model + alpha1*base_model1 + alpha2*base_model2 + ...) 

here, Base Model = 0, base_model_i = similarity_weight from DTi of selected record

Now construct a Decision Tree (Base model3) which takes same independent features but Final O/P upto previous DT value for training

Note: Got updated P, find the Residual2 R2 and repeat the above process again

#### Regression

Consider the following Dataset

Experience & Gap(Yes/No) (independent features) & Salary (dependent feature)

Base Model = mean(y)

Calculate Residual1 R1 = y - y_pred

Now create Base Model1 => DT with binary classifier

But in our dataset one of the independent feature (Experience) is a continuous feature

How do we create with Continuous feature as rootNode?

Take each value of Exp and check Eg. first record has Exp value as 2 => check for Exp<=2 on leftNode and Exp>2 on rightNode

Now calculate the Similarity weight

**Similarity Weight** = sum(Residual)** 2 / (no. of Residual + Lambda)

Eg. let's say we only have 1 residual that come under Exp<=2 category, then no. of residual = 1

Lambda is a hyperparameter, which on increasing decreases the similarity weight

Now calculate the Gain and the one which gives max Gain is selected for splitting (Eg; let's say Exp<=4 and Exp>4 has Highest Gain, choose that)

**Final O/P** = Base Model + alpha1*base_model1 + alpha2*base_model2 + ...

here, Base Model = 0, base_model_i = mean(residuals_i) for that node

Now construct a DT with binary classifier (Base model2) which takes same independent features but Final O/P upto previous DT value for training

##### Gamma hyperparameter

Used for Post Pruning to prevent from Overfitting

If Gain < Post Pruning, stop splitting that node
